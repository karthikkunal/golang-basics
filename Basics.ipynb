# What is Go ?

- Go (or Golang) is a statically typed, compiled programming language designed for simplicity,
- efficiency, and ease of use, with a focus on concurrent programming and scalability.
- Developed by Google, it has gained popularity for its performance and straightforward syntax.

**Go has 25 keywords.**
- This simplicity is intentional, reflecting Go's design philosophy of providing a minimalistic and readable syntax.
